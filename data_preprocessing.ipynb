### Aggregating data

Now, most csv files include timestamps which stop me from having one row per id, day_in_study, and study_interval. To address that, I will aggregate the data depending on their type for each csv file.

**Features and data type**

*wrist_temperature:*
I am using "temperature_diff_from_baseline" which is a level so the aggregation method will be mean, std, min, and max.

*heart_rate_variability_details:*
I am using "rmssd" and "high_frequency" which are a levels so the aggregation method will be mean, std, min, and max. I inlcuded std because if there's variation in HRV throughout the day, that can be a strong indicative of stress.

*resting_heart_rate:*
I am using "value" but the file has no timestamp column, it's a row per study_day so non-applicable.

*demographic_vo2_max:*
I am using "filtered_demographic_vo2_max" but the file has no timestamp column, it's a row per study_day so non-applicable.

*exercise:*
I am using "activeduration" which is a duration so the aggregation method will be summation.

*hormones_and_selfreport:*
I am using "phase" but the file has no timestamp column, it's a row per study_day so non-applicable.

```python
key = ['id', 'study_interval', 'day_in_study']
temperature_daily = (
    temperature.groupby(key)
    .agg(
          mean_temp=("temperature_diff_from_baseline", "mean"),
          max_temp=("temperature_diff_from_baseline", "max"),
          min_temp=("temperature_diff_from_baseline", "min"),
          std_temp=("temperature_diff_from_baseline", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(temperature_daily))
print("Key dupes:", temperature_daily.duplicated(key).sum())
temperature_daily.head()
```

So now what about the days where there was only one row to begin with. Now the std column will have NaN. This will be addressed in the "handling missing data" section.

But why do I even have std? The interesting result isn't "std predicts stress", it's whether it predicts stress more in the luteal phase than the follicular phase. I get at this by looking at feature importance within each phase separately, or with phase x feature interaction terms.

****std is a candidate feature. rmssd_std, for instance, measures how much a person's HRV swung across the day — a plausible stress signal that's distinct from the daily average. You don't "use" it manually; it goes into your feature matrix, and the feature-importance/SHAP step later tells you whether it actually predicted stress.****

Also, I am including max and min because for example, a day that's calm except for one sharp temperature spike has a high max but a smooth std and this can be captured and interpreted.

Here is the same code for the other files:

```python
hrv_daily = (
    hrv.groupby(key)
    .agg(
          mean_rmssd=("rmssd", "mean"),
          max_rmssd=("rmssd", "max"),
          min_rmssd=("rmssd", "min"),
          std_rmssd=("rmssd", "std"),

          mean_high_frequency=("high_frequency", "mean"),
          max_high_frequency=("high_frequency", "max"),
          min_high_frequency=("high_frequency", "min"),
          std_high_frequency=("high_frequency", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(hrv_daily))
print("Key dupes:", hrv_daily.duplicated(key).sum())
hrv_daily.head()
```

I am renaming the start_day_in_study column under exercise.csv to day_in_study so it matches the other files. This column does have the same purpose so renaming it is not an issue

```python
exercise = exercise.rename(columns={'start_day_in_study': 'day_in_study'})
```

```python

exercise_daily = (
    exercise.groupby(key)
    .agg(
          mean_activeduration=("activeduration", "mean"),
          max_activeduration=("activeduration", "max"),
          min_activeduration=("activeduration", "min"),
          std_activeduration=("activeduration", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(exercise_daily))
print("Key dupes:", exercise_daily.duplicated(key).sum())
exercise_daily.head()
```

### Merging everything onto anchor

Since files have different numbers of rows I decided to left-join on **phases** and use it as an anchor.

Before merging I wanted to check if there were any duplicate rows in the vo2, resting_hr and phases files which didn't need aggregation and therefore didn't have a duplicate check.

```python
key = ['id', 'study_interval', 'day_in_study']

for name, f in [('temperature_daily', temperature_daily),
                ('hrv_daily', hrv_daily),
                ('exercise_daily', exercise_daily),
                ('vo2', vo2),
                ('resting_hr', resting_hr),
                ('phases', phases)]:
    d = f.duplicated(key).sum()
    print(f"{name:20s} dupes on key: {d}")
```

There are lots of duplicates in the vo2 and resting_hr files. To fix this I will do the same aggregation steps.

```python
resting_hr_daily = (
    resting_hr.groupby(key)
    .agg(
          mean_bpm=("value", "mean"),
          max_bpm=("value", "max"),
          min_bpm=("value", "min"),
          std_bpm=("value", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(resting_hr_daily))
print("Key dupes:", resting_hr_daily.duplicated(key).sum())
resting_hr_daily.head()
```

```python
vo2_daily = (
    vo2.groupby(key)
    .agg(
          mean_vo2=("filtered_demographic_vo2_max", "mean"),
          max_vo2=("filtered_demographic_vo2_max", "max"),
          min_vo2=("filtered_demographic_vo2_max", "min"),
          std_vo2=("filtered_demographic_vo2_max", "std"),
    )
    .reset_index()
)

print("Daily rows:", len(vo2_daily))
print("Key dupes:", vo2_daily.duplicated(key).sum())
vo2_daily.head()
```

```python
key = ['id', 'study_interval', 'day_in_study']

for name, f in [('temperature_daily', temperature_daily),
                ('hrv_daily', hrv_daily),
                ('exercise_daily', exercise_daily),
                ('vo2_daily', vo2_daily),
                ('resting_hr_daily', resting_hr_daily),
                ('phases', phases)]:
    d = f.duplicated(key).sum()
    print(f"{name:20s} dupes on key: {d}")
```

I want the phases variable to only be the file with the column phases, excluding the other columns.

```python
key = ['id', 'study_interval', 'day_in_study']

phases = phases[key + ['phase']]
```

```python
anchor = phases
anchor_rows = len(phases)

merged = (phases
          .merge(temperature_daily, on=key, how='left', validate='one_to_one')
          .merge(hrv_daily,         on=key, how='left', validate='one_to_one')
          .merge(exercise_daily,    on=key, how='left', validate='one_to_one')
          .merge(vo2_daily,    on=key, how='left', validate='one_to_one')
          .merge(resting_hr_daily,    on=key, how='left', validate='one_to_one')
          )
```

Checking that the anchoring didn't change the number of rows that phases had. If they're equal, we're good!

### Handling Missing Data

Since we ensured data fell within a specific range and aggregated the data, there are no missing values within the _daily files. We can see that only std has NaN values and that's because if std is undefined for a single data point, it measures spread across multiple readings, so a day with only one reading has no spread to compute. These NaNs therefore indicate days where only one measurement was recorded, not missing data in the usual sense.

To deal with that, and because we want to maintain the std column, we fill the NaN values with 0. Filling with 0 reflects no observed spread while preserving the day's other valid statistics (mean, min, max) rather than dropping the row and losing data.

For the one row with NaN in phases, I removed it.

See below

```python
print(temperature_daily.isna().sum())
print(hrv_daily.isna().sum())
print(resting_hr_daily.isna().sum())
print(vo2_daily.isna().sum())
print(exercise_daily.isna().sum())
print(phases.isna().sum())
```

```python
# seeing the row with a missing phase
print(phases[phases['phase'].isna()])
```

```python
# removing the row with missing phase then remerging
phases = phases.dropna(subset=['phase'])

merged = (phases
          .merge(temperature_daily, on=key, how='left', validate='one_to_one')
          .merge(hrv_daily,         on=key, how='left', validate='one_to_one')
          .merge(exercise_daily,    on=key, how='left', validate='one_to_one')
          .merge(vo2_daily,    on=key, how='left', validate='one_to_one')
          .merge(resting_hr_daily,    on=key, how='left', validate='one_to_one')
          )
```

As we can see there are lots of NaN values and that is a result of the left join. To fix these NaN values I will look at the percentage of missing values and based on that, decide what method is most appropriate.

```python
print((merged.isna().mean()*100).round(1)[lambda s: s > 0].sort_values(ascending=False))
```

Since the active_duration columns all have an extremely high missing percentage, I will drop them.

```python
# dropping the active_duration columns
merged = merged.drop(columns=['mean_activeduration', 'max_activeduration', 'min_activeduration', 'std_activeduration'])
```

For the rest I will use imputation but I need to first decide whether I will use statistical methods or time-series depending on the gap of missing values. **By gap I mean the typical longest-gap across participant-waves**.

Let's check temperature first:

```python
# this code finds the longest run of consecutive missing days per participant
def max_gap(s):
    return s.isna().astype(int).groupby(s.notna().cumsum()).sum().max()

print(merged.groupby(['id','study_interval'])['mean_temp'].apply(max_gap).describe())
```

For temperature, most gaps are short approx 2 days, but the tail is 32 which is very long meaning some participants have a very long stretch of not having any recordings. For the short gaps, I will use time-series imputation.

I will use interpolation instead of forward-fill because temperature is a signal that changes gradually and continiously and I can capture that through interpolation.

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

merged['mean_temp'] = (merged
    .groupby(['id', 'study_interval'])['mean_temp']
    .transform(lambda s: s.interpolate(method='linear', limit=3, limit_area='inside')))

# filling the rest with the median

merged['mean_temp'] = merged['mean_temp'].fillna(merged['mean_temp'].median())
```

I will run the same process for the rest of the columns in my data and write out which method I used and why after the code.

```python
cols_to_check = ['max_temp', 'min_temp', 'mean_rmssd', 'min_rmssd', 'max_rmssd', 'mean_high_frequency', 'min_high_frequency', 'max_high_frequency',
                 'mean_bpm', 'min_bpm', 'max_bpm', 'mean_vo2', 'min_vo2', 'max_vo2']

for col in cols_to_check:
    gaps = merged.groupby(['id', 'study_interval'])[col].apply(max_gap)
    print(f"{col}: mean gap = {gaps.mean():.1f}")
```

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

merged['min_temp'] = (merged
    .groupby(['id', 'study_interval'])['min_temp']
    .transform(lambda s: s.interpolate(method='linear', limit=3, limit_area='inside')))

# filling the rest with the median

merged['min_temp'] = merged['min_temp'].fillna(merged['min_temp'].median())
```

```python
key_sort = ['id', 'study_interval', 'day_in_study']
merged = merged.sort_values(key_sort)

merged['max_temp'] = (merged
    .groupby(['id', 'study_interval'])['max_temp']
    .transform(lambda s: s.interpolate(method='linear', limit=3, limit_area='inside')))

# filling the rest with the median

merged['max_temp'] = merged['max_temp'].fillna(merged['max_temp'].median())
```

```python
median_cols = [
    'mean_vo2', 'max_vo2', 'min_vo2',
    'mean_rmssd', 'max_rmssd', 'min_rmssd',
    'mean_high_frequency', 'max_high_frequency', 'min_high_frequency',
    'mean_bpm', 'max_bpm', 'min_bpm'
]

for col in median_cols:
    merged[col] = merged[col].fillna(merged[col].median())
```

```python
# confirming these columns have no NaNs
print(merged[median_cols].isna().sum())
```

**Summary of other columns and how I handled missing data**

All the other columns had missing data but not enough to drop so I used imputation with them to depending on their gap.

For min_temp and max_temp I used the same process I did with mean_temp because the gap was still about 3.6 which is pretty small but there were still larger gaps.

For Vo2, although the gap was also small, about 4.2, Vo2 itself is a slow measurement that barely changes so interpolation won't do much, that's why I used median. I could've also used forward-fill (which keeps measurement constant) but it would've been the same as using median.

For the rest, I used median as the gaps were pretty large (e.g. 11.6 days)
